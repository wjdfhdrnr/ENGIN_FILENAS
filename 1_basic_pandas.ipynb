```python
df = pd.read_csv("test.csv")

df
```

```python
# 1. column 추가
df['ID'] = [1, 2, 3, 4, 5]
df
```

```python
# 2. column 삭제
df_2 = df.drop('ID', axis=1, inplace=False) # axis=1은 column을 의미합니다. inplace=True는 원본을 변경하겠다는 의미입니다.
df_2
df
```

```python
# 3. row 추가
df.loc[5] = ['F', 60, 600, 6]
df
```

```python
# 4. row 삭제
df.drop(5, axis=0, inplace=True) # axis=0은 row를 의미합니다.
df
```

```python
# 5. column 값 변경
df['age'] = df['age'] + 10
df
```

```python
# 6. row 값 변경
df.loc[0] = ['A', 10, 100, 2]
df
```

```python
# ## 그럼 Dataframe의 특정 행, 열의 값을 바꿀 수도 있을까요?
# Hint는 특정 행렬 값을 출력하는 코드를 참고하시는 것입니다.

df.iloc[0, 1] = 20
df
```

```python
df = pd.DataFrame({
    'name': ['노서준', '이채원', '김정민', '진서율', '박연서'],
    'age': [25, 23, 21, 26, 24],
    'score': [90, 89, 94, 85, 87],
    'ID': [123, 111, 134, 115, 147],
    'gender': ['남자','여자', '여자', '여자', '남자']
})
df
```

```python
# 7. column 순서 변경
df = df[["ID", "name", "age", "score", "gender"]]
df
```

```python
# 8. index 변경
df.set_index('ID', inplace=True)
df
```

```python
# 9. sorting
# sort by index
df.sort_index()
# sort by column
df.sort_values(by='age', ascending=False)
```

```python
# 10. banding을 통한 column 값 변경
df['grade'] = df['score'].apply(lambda x: 'A' if x >= 90 else 'B' if x >= 80 else 'C' if x >= 70 else 'D' if x >= 60 else 'F')
df
```

## 다양한 통계치 함수
### count : 전체 row의 개수를 반환합니다.
### mean : 평균을 반환합니다.
### std : 표준편차를 반환합니다.
### min : 최소값을 반환합니다.
### max : 최대값을 반환합니다.
### 25%, 50%, 75% : 백분위수를 반환합니다.

```python
df['age'].count() # age column의 전체 row의 개수를 반환합니다.
df['age'].mean() # age column의 평균을 반환합니다.
df['age'].std() # age column의 표준편차를 반환합니다.
df['age'].min() # age column의 최소값을 반환합니다.
df['age'].max() # age column의 최대값을 반환합니다.
df['age'].quantile(0.25) # age column의 25% 백분위수를 반환합니다.
df['age'].quantile(0.5) # age column의 50% 백분위수를 반환합니다.
df['age'].quantile(0.75) # age column의 75% 백분위수를 반환합니다.

# 이처럼 통계 함수를 통해 데이터의 전반적인 특징을 살펴볼 수 있습니다!
# 근데 이 방법은 너무 포괄적이라 세부적인 통계치를 확인하기 어렵습니다. 그래서 다음에는 groupby를 통해 세부적인 통계치를 확인하는 방법을 알아보겠습니다.
```

## Groupby
### Groupby는 SQL의 groupby와 유사한 기능을 제공합니다.
### 특정 column을 기준으로 그룹을 지어 그룹별로 통계치를 확인할 수 있습니다.
### * groupby 함수는 단독으로 쓰이기 보단 뒤에 통계치를 확인하는 다양한 함수와 함께 쓰입니다! *
### df.groupby('column_name').function() << 이런식으로 쓰입니다!

```python
# 예를들어 설별을 기준으로 그룹을 평균을 확인해보려면 어떻게 하면 될까요?
grouped_gender = df.groupby('gender')[['score', 'age']].mean()

grouped_gender
```

```python
grouped_gender_grade = df.groupby(['gender', 'grade'])['score'].mean()

grouped_gender_grade
```
